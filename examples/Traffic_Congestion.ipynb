### Visualize Traffic Congestion levels

https://www.kaggle.com/datasets/hasibullahaman/traffic-prediction-dataset/data

```python
import sys
import os

sys.path.append(os.path.abspath(os.path.join(os.getcwd(), '..')))

from kitikiplot.core import KitikiPlot
from kitikiplot.genomics import linear, grid

import random
import pandas as pd

os.chdir("..")
```

```python
df= pd.read_csv( "./datasets/Traffic/Traffic.csv" )
df= df[ ["Time", "Day of the week", "Traffic Situation"] ]

df= df.drop_duplicates()

print( "Shape: ", df.shape )

df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df["Day of the week"].value_counts()
```

```python
temp_df= df[ df["Day of the week"] == "Monday" ].drop_duplicates( subset= "Time" )
temp_df
```

```python
ktk= KitikiPlot( 
                    data= temp_df["Traffic Situation"].values.tolist(), 
                    stride= 1, 
                    window_length= temp_df.shape[0] )
ktk.plot(
        figsize= (3, 20),
        cell_width= 2,
        cmap= {
                'low': "lime", 
                'normal': 'cornflowerblue', 
                "high": "orange", 
                "heavy": "red"
            },
        transpose= False,
        window_gap= 0,
        xlabel= "",
        ylabel= "Time",
        yticks_values= temp_df["Time"].values.tolist(),
        # xticks_rotation= 90,
        display_xticks= False,
        display_yticks= True,
        title= "Traffic Congestion Levels ",
        display_legend= True,
        legend_kwargs= {
                            "bbox_to_anchor": (1.01, 1), 
                            "loc":'upper left', 
                            "borderaxespad": 0.
                        },
        kitiki_cell_kwargs= {"linewidth": 0.5}
    )
```
